### NumPy 맛보기 : 숫자 묶음을 한 번에 계산하기
    - '모든 값에 한 번에 계산'을 NumPy 배열로 해 보고, 배열의 모양 값으 종류 조건으로 고르기를 익힌 뒤 , 글자와 숫자를 섞응면 생기는 한계를 확인한다.

```python
import pandas as pd
import numpy as np
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

print(amounts)
# print("한 달 합계:", sum(amounts), "원")
```

## NumPy 배열로 바꾸면 한번에

## 배열 계산 모양 종류
- 배열로 합계 평균 가장 큰 지출을 구하고, 배열의 모양(shape)과 값의 종류(dtype)를 확인하세요.

```python
print("합계:",arr.sum())
print("평균:",round(arr.mean()))
print("가장 큰 지출:",arr.max())
print("모양:",arr.shape," 값의 종류:",arr.dtype)
```

## 식비만 합계 내기 -> 번거롭다
- 지출마다 분류(식비 교통 문화 쇼핑)를 붙이는 리스트를 하나 더 만들고, 반복문으로 식비 합계를 구해 보세요.

```python
categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

food = 0
for c, a in zip(categories, amounts):
    if c == "식비":
        food += a
print("식비 합계:",food, "원")
```

## 표로 바꾸기 : DataFrame
- 날짜(day) 리스트를 하나 더 적고, 네 리스르를 딕셔너리로 묶어 pd.DataFrame 표를 만드세요. 이름은 spend(지출).

```python
print(len(shops),len(categories),len(amounts))

spend = pd.DataFrame({
    "day": [2,3,5,7,9,12,14,16,19,22,25,28],
    "shop": shops,
    "category": categories,
    "amount" : amounts,
})

spend
```

# 열 꺼내서 계산하기
- 열 하나를 꺼내 합계 평균 최댓값 최솟값을 구하고 대괄호 한 겹(한 줄)과 두 겹(표)의 차이를 안다.

## 열 하나 꺼내기
- 금액(amount) 열 하나만 꺼내 보세요.


```python
spend["amount"]
```

## 합계 평균 최대 최소
- 꺼낸 금액 열로 합계,평균, 가장 큰 지출, 가장 작은 지출을 구하세요.

```python
print("합계 :",spend["amount"].sum())
print("평균 :",round(spend["amount"].mean()))
print("가장 큰 지출:", spend["amount"].max())
print("가장 작은 지출:",spend["amount"].min())
```

# 조건으로 행 고르고 정렬하기
- 조건식으로 True/False 줄을 만들어 원하는 행만 고르고, 두 조건을 &로 묶고, 금액 순으로 정렬해 TOP 3를 뽑는다

## 조건을 물으면서 True / False 줄 
- 분류가 '문화'인지 행마다 물어보세요.
    

```python
spend["category"] == "문화"
```

## Ture 인 행만 고르기

```python
spend[spend["category"]=="문화"]
```

## 숫자 조건: 만원 이상
- 한 번에 10,000원 이상 쓴 지출만 골라 보세요.

```python
spend[spend["amount"] >= 10000]
```

## 두 조건 한꺼번에 -> 오류
- '식비이면서 5,000 원 이상'을 & (그리고)로 이어 붙여 보세요. 괄호 없이!

## 조건마다 괄호로 감싸기
- 두 조건을 각각 괄호로 감싸서 다시 실행하세요.

```python
spend[(spend["category"] == "식비") & (spend["amount"] >= 5000)]
```

## 금액 순으로 정렬
- 금액이 작은 것부터 큰 것 순서로 정렬해 보세요.


```python
spend.sort_values("amount")
```

## 가장 큰 지출 Top 3
- 큰 금액부터 정렬한 뒤 앞의 3줄만 뽑아 top3 에 담으세요.

```python
top3 =  spend.sort_values("amount", ascending=False).head(3)
top3

# ‘식비이면서 5,000원 이상’을 고르는 코드는?

spend[(spend["category"] == "식비") & (spend["amount"] >= 5000)]

# 금액이 큰 3건을 뽑는 코드는?

spend.sort_values("amount", ascending=False).head(3)

```

## 새 열 만들고 분류별로 묶기
- 계산으로 새 열(포인트)을 만들고, 잘못 적은 한 칸을 loc로 고치고, groupby 로 분류별 합계를 구한다.

### 새 열: 적립 포인트
- 체크카드가 100원마다 1포인트를 줍니다. 금액으로 계산한 point 열을 새로 만드세요.

```python
spend["point"] = spend["amount"] // 100
spend.head(4)
```

## 포인트 합계
- 이번 달 적립 포인트 합계를 구하세요.

```python
print("이번 달 적립 포인드:",spend["point"].sum())
```
