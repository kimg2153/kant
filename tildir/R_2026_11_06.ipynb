# 절차지향과 객체지향
    - 절차지향은 데이터와 함수가 따로 있고, 객체지향은 상태(데이터)와 행동(메서드)을 객체 하나로 묶습니다.
    - 지금까지 쓴 리스트도 객체예요.
    - fruits.append() 는 리스트 클래스의 메서드입니다.


```python
class User:
    def __init__(self, name):
        self.name = name
    def show(self):
        print(f"회원 이름: {self.name}")

user_instance = User("스파르타")
user_instance.show()
```

## 생성자 __init__

```python
class Character:
    def __init__(self, nickname):
        self.nickname = nickname

player = Character("스파르타")
print(player.nickname)
# Character()
```

## Self
- self 는 지금 메서드를 실행하고 있는 객체 자신입니다. screen.check_myself() 라고 부르면 파이썬이 screen 을 self 자리에 자동으로 넣어줘요.
- self 는 카워드가 아니라 관례적인 이름 입니다. 그래도 반드시 self 로 씁시다

```python
class Monitor:
    def check_myself(self):
        print(f"메서드 안의 self 주소 : {id(self)}")

screen = Monitor()
print(f"실제 변수의 주소: {id(screen)}")
screen.check_myself()

import keyword
print(keyword.iskeyword("self"))
```

## BankAccount 와 __str__ __repr__

- 은행 계좌 클래스에 입금 출금 메서드와 출력용 매직 메서드를 만듭니다. 클래스 셀은 싱행해도 출력이 없어요.

- print(객체) 는 __str__, 셀 마지막 줄의 객체와 리스트 안의 객에츤 __repr__ 이 쓰입니다.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def __str__(self):
        return f"{self.owner}님의 계좌 (현재 잔액: {self.balance}원)"
    def __repr__(self):
        return f"BankAccount(owner='{self.owner}', balance={self.balance})"
    def deposit(self, amount):
        self.balance += amount
        print(f"[{self.owner} 계좌] {amount}원 입금 / 현재 잔액: {self.balance}원")
        return self.balance
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"[{self.owner} 계좌] {amount}원 출금 / 현재 잔액: {self.balance}원")
        else:
            print(f"[{self.owner} 계좌] 출금 실패: 잔액이 부족합니다. (잔액: {self.balance}원)")
        return self.balance

account_chulsoo = BankAccount("김철수", 5000)
account_younghee = BankAccount("이영희", 10000)
print(account_chulsoo)
account_younghee
```

## 메서드 실행과 독립성
- 철수 계좌에 입금, 영희 계좌에서 출금하면 우연히 둘 다 7000원이 됩니다. 셀 끝의 7000은 마지막 줄 반환값으 자동 출석이에요
- 독립성은 한 계좌만 바꿔 보면 확인됩니다.

```python
account_chulsoo.deposit(2000)
account_younghee.withdraw(3000)

account_chulsoo.deposit(500)
print(account_chulsoo.balance, account_younghee.balance)
```
