## 딕셔너리 (Dictionaray)


- {"키": "값"}
- "키" 중복 불가
- "값" 중복 가능
- 추가하는 순서를 유지

`````python
{"a":1} 
{"a":1,"b":1} # 값 중복 가능
# print({"a":1,"b":2}) 
print({"a":1,"b":1}) 
`````

### 딕셔너리 주요 메서드

- keys(),values(),items()

`````python
menu = {"coffee" : 4000, "tea": 4500}

# Key 목록 목회
print(menu.keys()) 

# value 목록 조회
print(menu.values())

# itmes 조회 (키, 값) 튜플 형태로 리
print(menu.items()) 

# 딕셔너리는 인덱서 방식으로 값을 참조 불가 > List 로 변환
print(list(menu.keys())[1])

print(sum(menu.values()))
`````

`````python
print("coffee" in menu) #True
print(4000 in menu) #False

print(4000 in menu.values()) #True
`````

`````python
# menu["cake"] 메뉴에 cake가 없기 때문에 에러
price =  menu.get("cake", 5000) # 케이크가 없기 때문에 5000을 리턴
print(price) 

# default 값이 없는 경우
print(menu.get("cake")) # 케이크가 없는경우 리턴값을 안정했기 때문에 None 반환

print(menu.get("coffee",0)) # coffee 는 존재하기 때문에 4000을 리턴
print(menu)
`````

`````python
print("coffee" in menu) #True  커피가 메뉴 키에 있는지
print(4000 in menu) #False 4000이 메뉴 키에 있는지

print(4000 in menu.values()) #True 4000이 메뉴 값에 있는지 확인
`````

### 집합 (Set)


수학 집합 : 중복 없고, 순서 없는 자료형

- 중괄호 사용
- set()함수를 생성


`````python
grade = set([1, 2, 2, 3, 3, 3]) # 집합을 생성 중복이 없기 때문에 {1,2,3}

print (grade)
`````

`````python
print (type({})) # {} 빈 딕셔너리
print (type(set())) # set() 빈 집합
print ({1,2,2}) # 집합
`````

`````python
s = {1,2} # 집합 선언

s.add(2) # s에 2를 추가하지만 집합이라 그냥 12
s.add(5) # s에 5를 추가해서 {1,2,5}

print(s)

s.remove(1) # s에 1을 제거해서 {2,5}

print(s)

s.discard(20) # 없는 키 삭제할 때도 오류가 없음
print(s)
`````

`````python
set_a = {1,2,3}
set_b = {3,4,5}

print(set_a & set_b) # 교집합
print(set_a | set_b) # 합집합
print(set_a - set_b) # 차집합
print(3 in set_a) # 포함 여부
`````

### for 반복문

```python

for 변수 in 박복할_데이터:
    반복수행할 코드 #1
    반복수행할 코드 #2
````

`````python
# 리스트
fruits = ["사과","바나나","체리"]

for fruit in fruits:
    print(fruit)

for index,fruit in enumerate(fruits):  # 인덱스와 값 가지고 오기
    print(index, fruit)

for ch in "스파르타":  # 문자를 1개씩 출력
    print(ch)

menu = {"아아":4000,"홍차":4500}

menu.keys()
for key in menu:  # 메뉴의 키값을 출력
    print(key)


for name,price in menu.items():  # 각각 name 과 price에 저장해 출력
    print(f"{name}:{price}")

    
`````

### While 문

```python
while 조건식:
    반복할 로직 #1
    반복할 로직 #2
    반복할 로직 #3

```


`````python
num = 1

while num < 3:  
    print(f"현재 숫자는 {num}")
    num += 1  #반복할때마다 num 을 1씩 더해서 2가되면 종료
`````

`````python
count = 3

while count > 0:
    print(count)
    count -= 1 # count = count -1  -를 사용해서 역순으로 출력
`````

`````python
for num in range(1,6): 
    if num == 3: # num이 3이 되면 종료
        print("즉시 종료")
        break
    print(f"현재 num={num}")
`````

`````python
while True: # True라 무조건 반복
    user_input = input() # 사용자가 입력
    if user_input == "q":  # 입력값이 q일때만 종료
        print("종료")
        break
`````

`````python
for num in range(1,6):
    if num % 2 == 0: #짝수
        continue
    print(f"홀수 {num}")
`````

`````python
# 구구단


for x in range(2,10): # 2~ 9
    
    if x > 9:  # x 가 9를 초과하면 정지하도록
        break
    print(f"--- {x}단 ---")
    for y in range(1,10): # 1~9 까지 
        if y == 10:
            break
        print(f"{x} x {y} = {x * y}") 
`````
