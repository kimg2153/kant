```python
## 클랙스 정의(선언) : 붕어빵 틀
class User:
    def __init__(self,name): # 생성자 : 클래스가 처음 생성될 때 자동으로 호출
        self.name = name # 변수를 초기화


    def show(self): # 메서드 (Method)
        return f"회원 이름: {self.name}"

# 인스턴스 생성: 붕어빵 만들기
user_a = User("손흥민")
user_b = User("이강인")

# User 클래스로 생성한 인스턴스(메모리에 생성)

# 클래스의 인스턴스 변수
print(user_a.name)
print(user_b.name)

print(user_a.show)
print(user_a.show)


print (type(user_a))
print (user_a is user_b)
print (id(user_a))
print (id(user_b))

```

```python
class SmartPhone: 
    def __init__(self, model):
        self.model = model # 속성 (Attibute)
        self.battery = 50 # 속성 (Attibute)

    def charge(self, amount):
        self.battery += amount
        if self.battery > 100:
            self.battery = 100
            print(f"배터리 충전 완료 : {self.battery}%")
        else :
            print(f"배터리 충전 중 ...: {self.battery}%")
            

phone_a = SmartPhone("아이폰")
print(phone_a.model)
phone_a.charge(30)
phone_a.charge(30)

phone_b = SmartPhone("갤럭시")
print(phone_b.model)
phone_b.charge(30)
phone_b.charge(30)
```

```python
# 은행 계좌 클래스

class BankAccount: 
    def __init__(self, owner, balance):
        self.owner = owner          # 계좌 예금주
        self.balance = balance      # 잔액

    def __str__(self):
        return f"예금주 : {self.owner}, 잔액: {self.balance:,}"

    def __repr__(self):
        return f"예금주 : {self.owner}, 잔액: {self.balance:,}"

    

    # 입급 처리 메서드
    def deposit(self,amount):
        self.balance += amount
        print(f"{amount:,} 입금완료 / 잔액 : {self.balance:,}원")

    # 출금 매서드
    # 잔액 확인 => 잔액 부족 메시지 출력
    def withdraw(self,amount):
        # print(f"통장 잔액 : {self.balance:,}")
        # if self.balance >= amount :
        #     self.balance -= amount
        #     print(f"{amount:,} 출금완료 / 잔액 : {self.balance:,}원")
        # else :
        #     print(f"통장 잔액 : {self.balance:,} \n잔액이 부족합니다.")

        if self.balance < amount:
            print(f"잔액 부족")
            return 0

        self.balance -= amount
        print(f"{amount:,} 출금 처리")
        return self.balance
                # Psudo code 

        

account_a = BankAccount("김준", 100_000)
print(account_a)
account_a.deposit(500_000)
account_a.withdraw(1_000_000)
print(f"잔액:{account_a.withdraw(200_000):,}")
print(f"잔액:{account_a.withdraw(300_000):,}")


account_c = BankAccount("홍길동",200_000)
print([account_a,account_c])
```

```python
# shopping cart 클래그 선언
# 상품 목록 저장 기능
# 메서드
# - 장바구니 담기
# - 장바구니 빼기
# - 상품 목록 조회
class ShoppingCart:
    #저장 기능
    def __init__(self):
        self.bucket = []
    # 장바구니 담기
    def add_item(self,item):
        self.bucket.append(item)
    # 장바구니 빼기
    def remve_itme(self,item):
        self.bucket.remove(item)
    # 상품 목록 조회
    def show_items(self):
        return self.bucket
# 삼품 클래스
# 속성
# - 제품명
# - 가격
class Item :
    def __init__(self,name,price):
        self.name = name
        self.price = price
# 쇼핑 카트 생성

cart = ShoppingCart()
# 상품 3개 생성 
item_a = Item("초코파이",5000)
item_b = Item("새우깡",2000)
item_c= Item("빅파이",3000)



# 상품 1개  장바구니 담기
cart.add_item(item_a)
# 상품 1개  장바구니 담기
cart.add_item(item_b)
# 상품 1개 장바구니 담기
cart.add_item(item_b)
# 삼품 1개 장바구니에서 빼기
cart.remve_itme(item_b)
# 장바구니에 담긴 상품 목록 조회
for item in cart.show_items():
    print(item.name, item.price)
```
