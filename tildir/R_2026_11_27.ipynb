## 1. 선형 회귀(Linear Regression)와 오차(Error)
- 선형 회귀(Linear Regression)란?
> 원인(입력값)과 격ㄹ과(정답) 사이에 비례하는 규칙이 있다고 가정하고, 이를 가장 잘 나태내는 '직선'을 그어 미래를 예측하는 머신러닝 기법입니다.    

### 1.1 예측을 위한 마법의 공식 (직선의 방정식)
> 컴퓨터가 집값을 예측할 때 사용하는 공식은 중학교 수학시간에 배운 1차 함수와 모양이 같습니다.

\hat{y} = wx + b

- \hat{y} (예측값): 인공지능이 계산해 낸 '예상 집값'입니다.
- x (입력값): 예측의 힌트가 되는 데이터, 즉 '방의 크기(평수)'입니다.
- w (가중치): 방이 1평 커질 때마다 가격이 얼마나 오르는지를 나타내는 '기울기'입니다. 이 값이 클수록 집값에 미치는 영향이 큽니다.
- b (편향): 방 크기가 0이어도 기본적으로 존재하는 '기본값(예: 땅값)'입니다. 선을 위아래로 움직여주는 역할을 합니다.



> 결국 머신러닝에서 '학습'시킨다는 것은, 주어진 데이터를 바탕으로 컴퓨터가 오차를 최소화하는 최고의 $w$와 $b$를 스스로 찾아내는 과정입니다.


### 1.2 모델이 얼마나 틀렸을까? (오차와 손실 함수)
> 오차(Error)와 손실함수(Loss Funciton)
---
> 컴퓨터가 그은 선이 실제 집값과 얼마나 차이가 나는지 점수를 매겨야 합니다. 이 차이를 오차(Error)라고 하며, 오차들을 모아 모델의 성능을 평가하는 공식을 손실 함수(Loss Funciton)라고 부릅니다. <br>
그리고 이 손실 함수에 현재 모델의 예측값과 실제값을 넣어 계산해서 나온 구체적인 수치(점수)를 손실값(Loss)이라고 합니다. 모델의 학습을 진행하면서 이 손실값을 0에 가깝게 줄이는 것을 목표로 삼습니다.



### 평균 제곱 오차(MSE: Mean Squared Error)
- 각각의 오차를 '제곱'한 뒤 평균을 낸 값입니다.

MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2


> 오차를 제곱하기 때문에 예측이 크게 빗나간 이상치(예: 뜬금없이 너무 비싼 집)에 큰 벌점을 주는 경향이 있습니다. 모델이 큰 실수를 하지 않도록 엄격하게 훈련시킬 때 주로 사용합니다.

### 평균 절대 오차(MAE: Mean Absolute Error)
- 각각의 오차에 '절대값'을 씌워(부호를 떼고) 평균을 낸 값입니다.

MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|

> 우리가 일상에서 쓰느 돈의 단위(예:"평균적으로 2천만 원 정도 틀렸네!") 와 스케일이 똑같아서 결과를 해석하고 보고하기에 매우 직관적입니다.

### 그래프를 통해 할 수 있는 주요 차이점
- 이상치(Outlier)에 대한 민감도 - 오차가 클 때 ($|e| > 1$ 영역)

> 그래프를 보면 오차가 1보다 커질 때 MSE(파란색 실선)가 MAE(주황색 점선)보다 훨씩 가파르게 위로 솟구치는 것을 볼 수 있습니다.<br>
오차가 2일때, MAE의 손실값은 2이지만 MSE의 손실값은 2^2 = 4가 됩니다. 만약 오차가 10 이라면 MSE는 100이라는 거대한 패널티를 부여합니다.<br>
따라서 MSE는 큰 오차(이상치)에 매우 민감하게 반응하여 이를 강하게 억제하려고 하고, MAE는 오차의 크기에 비례해서 선형적으로만 증가하므로 이상치에 비교적둔감(Robust)합니다.
