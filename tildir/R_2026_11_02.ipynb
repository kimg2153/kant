### 논리 연산자
- and
- or
- not

````python
is_adult = True
has_ticket = False

# and 연산
print(is_adult and has_ticket) #False
# or 연산
print(is_adult or has_ticket) #True
# not 연산
print(not is_adult) #False


print( False and (10/0 >1)) #10/0으로 ZeroDivisionError 이지만 논리 연산자가 and라서 앞의 False 만으로 조건을 끝냄

# print(False or (10/0 >1)) # 앞이 거짓이라 뒤에 조건도 확인하게 되다니보니 ZeroDivisionError 에러 발생
````

### 문자열

````python
print("선발 출전: %s,%d" % ("손흥민",20)) #s 문자, d 숫자
print("선발 출전: [%5d]" % (20)) #5d 로 5 칸 띄고 20 작성
print("소숫점 표기: %.2f" % 3.141592) #.2로 소숫점 둘째자리까지 표기
````

````python
name = "Cake"
price =  6500 
qty = 2

#format 방식
format_str = "주문내역: {0} {1} -> 총 {2}원".format(name,qty,price)
print(format_str)

# f-string 방식 
format_str = f"주문내역: {name} {qty} -> 총 {(price * qty):>10,}원" # 앞에 f를 붙이고 :이후에 > 우측  < 좌측  정렬 10 칸 띄고 , 돈 단위 표시
print(format_str)
````

````python
# 가상의 주문 데이터
orders = [
    ("Latte", 5000 , 2),
    ("Americano", 3000, 1),
    ("Cake", 6000, 1)
]

print("=" * 36)
print(f"{'메뉴명':<10} {'단가':>8} {'금액':>11}")
print("-" * 36)

#총금액
total_amount = 0

for menu, price , qty in orders:
    subtotal = price + qty
    total_amount += subtotal
    print(f"{menu:<10}{price:>13}{subtotal:>13}")
print("=" * 36)
print(f"{'합계금액:':<10}{total_amount:>22}원")
````

### 인덱스
```python
colors = ["red","green","blue"]
```
- 정방향 : 0, 1, 2
- 역방향 : -3, -2, -1  

````python
colors = ["red","green","blue"]
print(colors[0]) # 0 번째 red 출력
print(colors[-1]) # 역방향이라 blue 출력
````

````python
numbers = [0,1,2,3,4,5]
# 1번 부터 4번 미만까지
numbers[1:4] # [1, 2, 3] 출력
# 2칸 간격으로 출력
numbers[::2] # [0, 2, 4] 출력
# 역순으로 출력
numbers[::-1] # [5, 4, 3, 2, 1, 0] 출력
# 3번째 부터 출력
numbers[3:] # [3, 4, 5] 출력
# 앞의 3개만 출력
numbers[:3] # [0, 1, 2] 출력
````

````python
print((10,20,30,40)[1:3]) # 1번 값부터 3번 미만 값까지 즉 2번 값까지 출력해서 (20, 30) 출력
````

````python
matrix = [
    [1,2,3],
    [4,5,6]
]

print(matrix[1]) # [4, 5, 6]
print(matrix[1][2]) # 6
print(len(matrix), len(matrix[0])) # 배열의 길이 2 , 0번 배열의 길이 3
````
